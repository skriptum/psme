## Basic RBC Model 

## Households

$$ u(c_t) + \nu(n_t) = ln c_t - \chi \frac{n_t^{1+\eta}}{1+ \eta}$$

#### Problem of the Household

\begin{align*}
 max_{\{c_t, n_t, i_t\}_{t=0}^{+ \infty}} & \qquad \mathbb{E}_0 \sum_{t = 0}^{+\infty} \beta^t (ln c_t - \chi \frac{n_t^{1+\eta}}{1+ \eta})\\
s.t. & \qquad c_t + i_t = w_t n_t + r_t k_t + \pi_t \qquad \forall t = 0,1,2, \dots \\
&\qquad k_{t+1} = (1 - \delta) k_t + i_t \qquad \forall t = 0,1,2, \dots
\end{align*}

We can go from the two contraints to just one: 
\begin{align*}
 max_{\{c_t, n_t, k_{t+1}\}_{t=0}^{+ \infty}} & \qquad \mathbb{E}_0 \sum_{t = 0}^{+\infty} \beta^t (ln c_t - \chi \frac{n_t^{1+\eta}}{1+ \eta})\\
s.t. & \qquad c_t + k_{t+1} = w_t n_t + r_t k_t + (1-\delta) k_t + \pi_t \qquad \forall t = 0,1,2, \dots \\
\end{align*}

In order to solve it, we can use the Lagrangian: 
$$ \mathcal{L} = \mathbb{E}_0 \sum_{t = 0}^{+\infty} \beta^t \{(ln c_t - \chi \frac{n_t^{1+\eta}}{1+ \eta}) + \lambda_t[w_t n_t + r_t k_t  + (1-\delta) k_t + \pi_t - c_t - k_{t+1}]\} $$

\begin{align*}
    \frac{\partial \mathcal{L}}{\partial c_t} &= \frac{1}{c_t} - \lambda_t = 0\\
    \frac{\partial \mathcal{L}}{\partial n_t} &= -\chi n_t^\eta + \lambda_t w_t = 0\\
    \frac{\partial \mathcal{L}}{\partial k_{t+1}} &= - \lambda_{t} + \beta \mathbb{E}_t[\lambda_{t+1}(r_{t+1} + (1-\delta))] = 0\\
    \frac{\partial \mathcal{L}}{\partial \\\lambda_{t}} &= -c_t -i_t + w_t n_t + r_t k_t + \pi_t = 0
\end{align*}

We can write this conditions as: 
\begin{align*}
    1 &= \mathbb{E}_t [\beta \frac{c_{t}}{c_{t+1}}(r_{t+1} + (1-\delta))] \\
    \chi n_t^\eta c_t &= w_t
\end{align*}



#### Transversality condition
$$
\lim_{T \to \infty} \mathbb{E}_0\left[ \beta^T u_c(c_T)\, k_{T+1} \right] = 0
$$

$$
\lim_{T \to \infty} \mathbb{E}_0\left[ \beta^T \frac{k_{T+1}}{c_T} \right] = 0
$$

## Firms

$$ y_t = z_tk_t^\alpha n_t^{1-\alpha}$$

#### Problem of the Firm

$$ \max_{k_t,n_t}  \pi_t = z_tk_t^\alpha n_t^{1-\alpha} -r_t k_t - w_t n_t$$

This is an easy problem which leads to: 
$$ r_t = \alpha z_t k_t^{\alpha-1} n_t^{1-\alpha}$$

$$ w_t = (1-\alpha) z_t k_t^{\alpha} n_t^{-\alpha}$$

## Law of Motions
- Technology evolves following an $AR(1)$ process with persistence $\rho$ $$ln z_{t} = \rho ln z_{t-1} + \varepsilon_t $$ where $\varepsilon_t \sim \mathcal{N}(0, \sigma_\varepsilon^2)$

- Capital law of motion is given by: $$k_{t+1} = k_t - \delta k_t + i_t$$ where $\delta$ is the rate of depreciation

## Equilibrium conditions
- ##### Consumption - savings (Euler Equation)
$$ 1 = \mathbb{E}_t [\beta \frac{c_{t}}{c_{t+1}}(r_{t+1} + (1-\delta))] $$
- ##### Consumption labour condition
$$\chi n_t^\eta c_t = w_t$$
- ##### Firms Capital and Labour demand
$$ \alpha z_t k_t^{\alpha-1} n_t^{1-\alpha} =  r_t$$

$$ (1-\alpha) z_t k_t^{\alpha-1} n_t^{-\alpha} = w_t$$

- #### Resource Constraint
$$ c_t + k_{t+1} - (1-\delta)k_t = z_t k_t^\alpha n_t^{1-\alpha} $$

- #### Law of Motion of Capital
$$ k_{t+1} = (1-\delta)k_t + i_t$$

- #### Law of Motion of Technology

$$ ln z_{t} = \rho ln z_{t-1} + \varepsilon_t $$

Just note that to be precise we are already using three conditions in one. In fact we should state the resource constaint as: 
$$ y_t = z_t k_t^\alpha n_t^{1-\alpha} $$
$$ c_t + i_t = y_t $$

Until here we have also skipped the labour market equilibrium condition in which labour demand $l_t$ is equal to labour supply $n_t$, assuming that this is always true. 

## Steady State

In the steady state: 
- No shocks happens : $ \varepsilon = 0$.
- All the variable are constant - no more $t+1, t, t-1$ indices.

Let's solve it in this case:
- #### Law of Motion of Capital
$$k_{ss} = (1-\delta)k_{SS} + i_{SS} \iff i_{SS} = \delta k_{SS} $$

- #### Technology
$$ ln z_{SS} = \rho ln z_{SS} + \varepsilon_t \iff ln z_{SS} = \rho ln z_{SS} \iff z_{SS} = 1$$

- #### Consumption - Labour Choice
$$ \chi n_{SS}^\eta c_{SS} = w_{SS}$$

- #### Euler Equation
$$ 1 = \beta (r_{SS} + 1 -\delta)$$

- #### Firms' Capital and Labour Optimality
$$ \alpha k_{SS}^{\alpha -1} n_{SS}^{1-\alpha} = r_{SS}$$
$$ (1-\alpha) k_{SS}^\alpha n_{SS}^{-\alpha} = w_{SS}$$

- #### Resource Constraint 
$$c_{SS} + \delta k_{SS} = k_{SS}^\alpha n_{SS}^{1-\alpha} \Longrightarrow c_{SS} = (\frac{r_{SS}}{\alpha} - \delta)k_{SS}$$

## Log-Linearization of equations

We will use small letters with an hat, as $\hat{x}_t$ to indicate log-linearized variables:

- Consumption - Savings (Euler Equation)
$$ \hat{c}_t = - \beta r_{SS} \mathbb{E}_t\hat{r}_{t+1} + \mathbb{E}_{t} \hat{c}_{t+1} $$

- Consumption - Labour 
$$ \hat{w}_t = \hat{c}_t + \eta \hat{n}_t$$

- Firms' Capital and Labour Optimality
$$ \hat{r}_t = \hat{z}_t - (1 - \alpha)(\hat{k}_t - \hat{n}_t)$$
$$ \hat{w}_t = \hat{z}_t +  \alpha(\hat{k}_t - \hat{n}_t)$$

- Law of Motion of Capital
$$ \hat{k}_{t+1} = (1-\delta) k_t + \delta \hat{i}_t$$

- Law of Motion - Technology 
$$\hat{z}_t = \rho\hat{z}_{t-1} + \varepsilon_t$$

- Production Function
$$ \hat{y}_t = \hat{z}_t + \alpha \hat{k}_t + (1-\alpha) \hat{n}_t$$

- GDP

$$ \frac{c_{SS}}{y_{SS}} \hat{c}_t + \frac{i_{SS}}{y_{SS}}\hat{i}_t = \hat{y}_t $$

Rewriting them together: 
$$ \hat{c}_t = - \beta r_{SS} \mathbb{E}_t\hat{r}_{t+1} + \mathbb{E}_{t} \hat{c}_{t+1} $$
$$ \hat{w}_t = \hat{c}_t + \eta \hat{n}_t$$
$$ \hat{r}_t = \hat{z}_t - (1 - \alpha)(\hat{k}_t - \hat{n}_t)$$
$$ \hat{w}_t = \hat{z}_t +  \alpha(\hat{k}_t - \hat{n}_t)$$
$$ \hat{k}_{t+1} = (1-\delta) k_t + \delta \hat{i}_t$$
$$\hat{z}_t = \rho\hat{z}_{t-1} + \varepsilon_t$$
$$ \hat{y}_t = \hat{z}_t + \alpha \hat{k}_t + (1-\alpha) \hat{n}_t$$
$$ \frac{c_{SS}}{y_{SS}} \hat{c}_t + \frac{i_{SS}}{y_{SS}}\hat{i}_t = \hat{y}_t $$

The system of equation we want to solve is the following:

\begin{align*}
0 &= -\hat{c}_t + \hat{z}_t +\alpha \hat{k}_t -(\eta + \alpha)\hat{n}_t  \\
-\beta r_{SS} (1-\alpha)\hat{k}_{t+1} + \beta r_{SS}(1-\alpha) \hat{n}_{t+1} + \beta r_{SS} \hat{z}_{t+1} - \hat{c}_{t+1} &= -\hat{c}_t \\
\frac{k_{SS}}{y_{SS}} \hat{k}_{t+1} &= \hat{z}_t + \frac{k_{SS}}{y_{SS}\beta} \hat{k}_t + (1-\alpha)\hat{n}_t - \frac{c_{SS}}{y_{SS}}\hat{c}_t
\end{align*}

```python
# importing packages

import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols
from scipy.optimize import fsolve
```

```python
# initialize parameters

α = 0.33
γ = 1
β = 0.99
ρ = 0.8
δ = 0.025
η = 1
χ = 7.88
σ_ε = 0.1
```

```python
# steady state values

Z_ss = 1
N_ss = 1/3 #assumed
K_ss = ((1-β*(1-δ))/(β*α))**(1/(α-1))*N_ss 
r_ss = α*K_ss**(α-1)*N_ss**(1-α)
w_ss = (1-α)*K_ss**α * N_ss**(-α)
I_ss = δ * K_ss                      
Y_ss = ((K_ss)**α)*((N_ss)**(1-α))
C_ss = Y_ss - δ*K_ss
```

```python
# solve the model in log-deviations

# Simulation of 1-time shock
T = 200

MIT_shock = np.zeros(T)
MIT_shock[0] = σ_ε
for t in range(1,T):
    MIT_shock[t] = ρ*MIT_shock[t-1]

Z_trans = np.zeros(T)
Z_trans = MIT_shock
```

```python
# Initial guess for x (size 1000x3)
x0 = np.array([0, 0, 0]).reshape(-1, 1) @ np.ones((1, T))  # This will be T x 3
x0 = x0.flatten()  # Flatten into a 1D array (fsolve expects a 1D initial guess)

def rbc_solver_log_lin(x, α, β, η, r_ss, K_ss,Y_ss,C_ss, Z_trans, T):

    k = x[:T]
    c = x[T:2*T]
    n = x[2*T:3*T]

    errors = np.zeros(3 * T)         

    # error in the first period for capital                  
    errors[T - 1] = k[0] - 0
    # error in the last period for capital
    errors[2 * T - 1] = k[T-1] - 0

    # error in the Euler equation
    for t in range(T - 1):
        errors[t] = c[t] - c[t+1] - β*r_ss*(1-α)*k[t+1] + β*r_ss*(1-α)*n[t+1] + β*r_ss*Z_trans[t+1]

    # error in the budget constraint
    for t in range(T - 1):
        errors[T + t] = K_ss/Y_ss*k[t+1] - Z_trans[t] - K_ss/(Y_ss*β)*k[t] - (1-α)*n[t] + C_ss/Y_ss*c[t]

    # error in the labor supply
    for t in range(T):
        errors[2 * T + t] = - c[t] + Z_trans[t] + α * k[t] - (η + α) * n[t]

    return errors

# Solve using fsolve
sol_log_lin = fsolve(rbc_solver_log_lin, x0, args=(α, β, η, r_ss, K_ss,Y_ss,C_ss, Z_trans, T), xtol=1e-8, maxfev=10000)

# Reshape the solution back to (T, 3)

print(sol_log_lin)
```

```python
K_trans_log = sol_log_lin[:T]
C_trans_log = sol_log_lin[T:2*T]
N_trans_log = sol_log_lin[2*T:]

Y_trans_log = Z_trans + α*K_trans_log + (1-α)*N_trans_log
I_trans_log = (K_trans_log[1:] - (1-δ)*K_trans_log[:-1])/δ
w_trans_log = Z_trans + α*(K_trans_log - N_trans_log)
r_trans_log = Z_trans - (1-α)*(K_trans_log - N_trans_log)
```

```python
plt.figure(figsize=(12, 12))  # Adjust the figure size

# Subplot for Technology
plt.subplot(8, 1, 1)  # 6 rows, 1 column, 1st subplot
plt.plot(Z_trans, label='Technology')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Technlogy')
plt.legend()

# Subplot for Capital
plt.subplot(8, 1, 2)  # 6 rows, 1 column, 2st subplot
plt.plot(K_trans_log, label='Capital')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Capital')
plt.legend()

# Subplot for Consumption
plt.subplot(8, 1, 3)  # 3 rows, 1 column, 3nd subplot
plt.plot(C_trans_log, label='Consumption')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Consumption')
plt.legend()

# Subplot for Labor
plt.subplot(8, 1, 4)  # 3 rows, 1 column, 4rth subplot
plt.plot(N_trans_log, label='Labor')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Labor')
plt.legend()

# Subplot for Output
plt.subplot(8, 1, 5)  # 5 rows, 1 column, 5th subplot
plt.plot(Y_trans_log, label='GDP')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('GDP')
plt.legend()

# Subplot for Investment
plt.subplot(8, 1, 6)  # 4 rows, 1 column, 6th subplot
plt.plot(I_trans_log, label='Investment')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Investment')
plt.legend()

# Subplot for Wage
plt.subplot(8, 1, 7)  # 7 rows, 1 column, 7th subplot
plt.plot(w_trans_log, label='Wage')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Wage')
plt.legend()        

# Subplot for Interest Rate
plt.subplot(8, 1, 8)  # 8 rows, 1
plt.plot(r_trans_log, label='Interest Rate')
plt.axhline(0, color='r', linestyle='--', label='Steady state')
plt.title('Interest Rate')
plt.legend()


# Adjust layout and display the plot
plt.tight_layout()  # Ensures there's no overlap between subplots
plt.show()
```
